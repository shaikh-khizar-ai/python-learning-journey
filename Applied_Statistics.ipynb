```python
# Applied Statistics - Day 01
# Freight Data Analytics

import pandas as pd

data = { "Shipment": ["SHIP001", "SHIP002", "SHIP003", "SHIP004", "SHIP005"],
        "Weight": [120, 85, 200, 150, 100],
        "Cost": [6000, 6800, 12000, 10500, 5500],
        "Delay": [2, -1, 5, 0, 3] }

df = pd.DataFrame(data)
df
```

```python
# Mean, Median, Mode
print(f"Average Weight: {df['Weight'].mean()}")
print(f"Average Cost: {df['Cost'].mean()}")
print(f"Average Delay: {df['Delay'].mean()}")

print(f"\nMedian Delay: {df['Delay'].median()}")
print(f"Mode Delay: {df['Delay'].mode()}")

print(f"\nStd Dev Delay:  {df['Delay'].std()}")
print(f"Variance Delay: {df['Delay'].var()}")
```

```python
# Statistics Day 2 - Range, Variance, Std Dev
import pandas as pd

# 30 Shipments Data
data = { "Shipment": ["SHIP001", "SHIP002", "SHIP003", "SHIP004", "SHIP005", "SHIP006", "SHIP007", "SHIP008", "SHIP009", "SHIP010", "SHIP011","SHIP012", "SHIP013", "SHIP014", "SHIP015", "SHIP016", "SHIP017", "SHIP018", "SHIP019", "SHIP020", "SHIP021", "SHIP022", "SHIP023", "SHIP024", "SHIP025", "SHIP026", "SHIP027", "SHIP028", "SHIP029", "SHIP030"],
        "Weight": [120, 85, 200, 150, 100, 250, 180, 90, 300, 110, 220, 160, 130, 75, 190, 280, 95, 210, 140, 170, 240, 105, 195, 80, 260, 175, 230, 145, 115, 205],
        "Cost": [6000, 6800, 12000, 10500, 5500, 22500, 11700, 6750, 25500, 6600, 20900, 11200, 7150, 6000, 12350, 25200, 7125, 14700, 7000, 10200, 22800, 6825, 13650, 6000, 15600, 15750, 16100, 10150, 6325, 13325],
        "Delay": [2, -1, 5, 0, 3, 7, 1, -2, 4, 0, 6, 2, -1, 3, 5, 8, 1, 4, 0, 2, 6, -3, 3, 1, 7, 4, 2, -1, 3, 5] }

df = pd.DataFrame(data)
df.head(10)
```

```python
# Range = Max - Min
weight_range = df["Weight"].max() - df["Weight"].min()
cost_range =  df["Cost"].max() - df["Cost"].min()
delay_range = df["Delay"].max() - df["Delay"].min()

print(f"Weight Range: {weight_range} kg")
print(f"Cost Range: ₹{cost_range}")
print(f"Delay Range: {delay_range} days")
```

```python
# Variance
print("=== Variance ===")
print(f"Weight: {df['Weight'].var():.2f}")
print(f"Cost: {df['Cost'].var():.2f}")
print(f"Delay: {df['Delay'].var():.2f}")
```

```python
# Standard Deviation
print("=== Standard Deviation ===")
print(f"Weight: {df['Weight'].std():.2f} kg")
print(f"Cost: ₹{df['Cost'].std():.2f}")
print(f"Delay: {df['Delay'].std():.2f} days")
```

```python
# Complete Summary
print("=" * 40)
print("FREIGHT DATA SUMMARY")
print("=" * 40)

for col in ["Weight", "Cost", "Delay"]:
    print(f"\n{col}:")
    print(f" Mean: {df[col].mean():.2f}")
    print(f" Median: {df[col].median():.2f}")
    print(f" Std: {df[col].std():.2f}")
    print(f" Range: {df[col].max() - df[col].min():.2f}")
```
