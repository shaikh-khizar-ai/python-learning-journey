```python
# Applied Statistics - Day 01
# Freight Data Analytics

import pandas as pd

data = { "Shipment": ["SHIP001", "SHIP002", "SHIP003", "SHIP004", "SHIP005"],
        "Weight": [120, 85, 200, 150, 100],
        "Cost": [6000, 6800, 12000, 10500, 5500],
        "Delay": [2, -1, 5, 0, 3] }

df = pd.DataFrame(data)
df
```

```python
# Mean, Median, Mode
print(f"Average Weight: {df['Weight'].mean()}")
print(f"Average Cost: {df['Cost'].mean()}")
print(f"Average Delay: {df['Delay'].mean()}")

print(f"\nMedian Delay: {df['Delay'].median()}")
print(f"Mode Delay: {df['Delay'].mode()}")

print(f"\nStd Dev Delay:  {df['Delay'].std()}")
print(f"Variance Delay: {df['Delay'].var()}")
```
